```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn import datasets
from sklearn.metrics import accuracy_score
```

```python
iris = datasets.load_iris()
X = iris.data[:, 0:2]
y = iris.target
```

```python
# Plot the ground truth
plt.figure(figsize=(8, 6))

sc = plt.scatter(X[:, 0], X[:, 1], c = y, cmap = 'tab10', edgecolor='k', s = 40, vmin=0, vmax=10)
plt.colorbar(sc)
plt.title('Ground Truth')
plt.show()
```

### Initialize the centers for kmeans

```python
estimators = [('k=9', KMeans(n_clusters=9)),#initialize the centroids by a smart way
              ('k=6', KMeans(n_clusters=6)),#initialize the centroids by a smart way
              ('k=3', KMeans(n_clusters=3)),#initialize the centroids by a smart way
              ('k=3(random initialization)', KMeans(n_clusters=3, n_init=1, init='random'))] #inistialize the centroids randomly
```

```python
for name, model in estimators:
    plt.figure(figsize=(8, 6))
    model.fit(X) #fit k-means for different number of clusters
    labels = model.labels_
    
    #let the color of the points related with the labels
    #define vmin and vmax for the upper and lower limits of the colorbar, such that all plots will have the same color scale, 
    #tab-10 is the colormap for discrete value
    sc = plt.scatter(X[:, 0], X[:, 1], c= labels, cmap = 'tab10', edgecolor='k', s = 40, vmin=0, vmax=10) 
   
    plt.colorbar(sc) #show colorbar
    
    #cluster_centers_ is an array, [row = n_clusters, col = n_features], 
    #the cluster centers are marked by 'x'
    plt.scatter(model.cluster_centers_[:, 0], model.cluster_centers_[:, 1], marker = 'x', s = 200, c = np.arange(len(model.cluster_centers_)), cmap = 'tab10', vmin=0, vmax=10) 
  
    plt.title(name)
    plt.show()
    print('The accuracy is: %f'%accuracy_score(y, labels)) #print the estimation accuracy
```
